```python
%%writefile ccalaculator.py
import pytest

def calculate(value):
    return 10 / int(value)

@pytest.mark.parametrize("value", [0, "abc", None])
def test_unknown_exception(value):

    with pytest.raises(Exception) as error:
        calculate(value)

    print("Exception Type:", error.type.__name__)
    print("Error Message:", error.value)
```

```python
!pytest -v -s ccalaculator.py
```

```python
%%writefile test_example2.py

def calculate(value):
    # Convert input to integer and divide 10 by it
    return 10 / int(value)


def test_valid_input():
    # Check if the function works with a valid input
    result = calculate("2")
    assert result == 5.0


def test_unknown_error():
    # We don't know which exception might occur
    result = calculate("abc")

    # Check expected result
    assert result == 10
```

```python
%%writefile test_example3.py

def calculate(value):
    # Convert input to integer and divide 10 by it
    return 10 / int(value)


def test_find_error():

    try:
        # Execute code that might generate an error
        calculate("abc")

    except Exception as e:
        # Print the type of exception
        print("Error name:", type(e).__name__)

        # Print details about the exception
        print("Error details:", str(e))
```

```python
!python -m pytest -v -s test_example3.py
```

```python
%%writefile calculator2.py
def divide(a,b):
  if b == 0:
    raise ValueError('b cannot be zero')
  return a / b
```

```python
%%writefile test_calculator2.py
import pytest
from calculator2 import divide

def test_divide_by_zero():
  with pytest.raises(ValueError) as exec_info:
    divide(10, 0)

  print(exec_info.value)
```

```python
!pytest -v -s test_calculator2.py
```
